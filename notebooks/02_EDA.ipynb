## Supplementary analysis: repetition, noise and fuel effects

Further checks that determine how the model should be validated and how much accuracy is attainable:

1. Duplicates on the columns the model actually uses (after the raw columns were dropped).
2. Same-vehicle repetition (trims / years), which motivates grouping by vehicle line in train/test splits and cross-validation.
3. Irreducible noise: identical specs, different CO2.
4. CO2 per L/100 km by fuel type.
5. `Gears` sanity (rare 1- and 4-speed rows, the `0 = CVT` encoding).
6. Rare `Make` levels.
7. Year trend and the bimodal CO2 distribution.

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='whitegrid')

clean = pd.read_csv('../datasets/cleaned_data/cleaned_cars_info.csv')
fe = pd.read_csv('../datasets/feature_engineered_data/feature_engineered_cars.csv')
assert len(clean) == len(fe)
fe['Model'] = clean['Model'].values            # row-aligned (verified by length; Model was dropped in NB3)
fe['Transmission'] = clean['Transmission'].values
fe = fe[fe['Fuel type'] != 'N'].reset_index(drop=True)   # single natural-gas row
T = 'CO2 emissions (g/km)'
FEATS = ['Model year','Make','Vehicle class','Engine size (L)','Fuel type','Transmission type','Gears']
print(fe.shape)
```

## 1. Duplicates on the columns the model actually sees

```python
print('Exact duplicate rows (features + target):', fe.duplicated(FEATS + [T]).sum(), 'of', len(fe))
print('Rows sharing a feature vector with another row:', fe.duplicated(FEATS, keep=False).sum(), 'of', len(fe))
print('Rows that are a repeat of an earlier feature vector:', fe.duplicated(FEATS).sum())
```

## 2. Same-vehicle repetition (model 'family' = Make + first word of Model)

```python
fe['Family'] = fe['Make'] + ' ' + fe['Model'].str.split().str[0]
fam = fe.groupby('Family').size()
print('Distinct families:', len(fam), '| median rows per family:', fam.median(), '| max:', fam.max())
print('Share of rows in families with >= 10 rows:', round(fam[fam>=10].sum()/len(fe),3))
print(fam.sort_values(ascending=False).head(10))
# Would a random 80/20 split put the same family on both sides?
rng = np.random.RandomState(42); mask = rng.rand(len(fe)) < 0.2
test_fams = set(fe.loc[mask,'Family']); train_fams = set(fe.loc[~mask,'Family'])
print('Share of test rows whose family also appears in train (random split):',
      round(fe.loc[mask,'Family'].isin(train_fams).mean(),3))
```

## 3. Irreducible noise: identical specs, different CO2

```python
g = fe.groupby(FEATS)[T].agg(['count','std','nunique'])
multi = g[g['count']>1]
print('Feature combinations:', len(g), '| repeated:', len(multi))
print('Repeated combos with >1 distinct CO2 value:', (multi['nunique']>1).sum())
print('Mean within-combo std (g/km):', round(multi['std'].mean(),2))
within = fe.groupby(FEATS)[T].transform(lambda s: s - s.mean())
print('Pooled within-combo RMSE floor (g/km), repeated rows only:',
      round(np.sqrt((within[fe.duplicated(FEATS, keep=False)]**2).mean()),2))
```

## 4. CO2 per L/100 km by fuel type

```python
fe['kg_per_100L'] = fe[T] / fe['Combined (L/100 km)']
print(fe.groupby('Fuel type')['kg_per_100L'].agg(['mean','std','count']).round(3))
sns.boxplot(data=fe, x='Fuel type', y='kg_per_100L'); plt.title('g CO2/km per (L/100 km), by fuel'); plt.show()
```

## 5. Gears sanity

```python
print(fe['Gears'].value_counts().sort_index())
print(fe[fe['Gears'].isin([1,4])].groupby(['Gears','Transmission type','Make']).size().head(25))
print(fe[fe['Gears']==0]['Transmission type'].value_counts())
```

## 6. Rare makes

```python
mk = fe['Make'].value_counts()
print('Makes:', len(mk), '| makes with < 30 rows:', (mk<30).sum())
print(mk[mk<30])
```

## 7. Year trend and the bimodal CO2 distribution

```python
print(fe.groupby('Model year')[T].agg(['mean','count']).round(1))
fe['Body'] = np.where(fe['Vehicle class'].str.contains('SUV|Sport utility|Pickup|Van|Minivan|Special'), 'SUV/Truck/Van', 'Car/Wagon/2-seater')
sns.kdeplot(data=fe, x=T, hue='Body', fill=True, common_norm=False); plt.title('CO2 by body group'); plt.show()
print(fe.groupby('Body')[T].describe().round(1))
```

## Findings (from the outputs above)

1. **Duplicates:** 1,196 rows are exact duplicates on features + target, and 4,686 rows repeat an earlier feature vector (7,189 rows share a vector with another row). On the raw 15 columns `Model`, City/Highway etc. keep rows distinct; the duplicates only appear once those columns are dropped.
2. **Same-vehicle repetition:** 443 model families, median 14 rows each; 92% of rows sit in families with >= 10 rows (Porsche 911: 311, F-150: 266). With a random 80/20 split, **99.7% of test rows have a family-mate in the training set**. Random-split metrics therefore measure interpolation between siblings, not generalisation to a new vehicle.
3. **Noise floor:** 2,179 of 2,503 repeated feature combos contain more than one CO2 value (mean within-combo std 9.3 g/km; pooled RMSE 12.4 g/km on those rows). The chosen features cannot separate those cars. This is a rough floor for repeated rows only, not for the whole dataset.
4. **Fuel type:** g CO2 per (L/100 km) is nearly constant within a fuel (std 0.2-0.4) but differs between fuels: X/Z about 23.4, D 26.9, E85 16.4. So fuel type changes the *slope* between consumption and CO2, not just the intercept. E85 has the lowest CO2 per litre, yet the highest mean CO2 in `02_EDA.ipynb`, which suggests E85 vehicles are mostly large, thirsty ones (not tested here).
5. **Gears:** `0` is used only for AV/CVT (410 rows), a sentinel on a numeric scale. 19 rows have Gears = 1 (all AV, Hyundai/Kia) and 21 have Gears = 4 (A/AS from Dodge, Nissan, Toyota, Scion): plausible but rare.
6. **Rare makes:** 4 of 42 makes have < 30 rows (smart 5, Ferrari 6, Scion 13, Bugatti 13).
7. **Year trend:** mean CO2 rises from 246.9 (2015) to about 260 (2022-24), mostly during 2020-2022. It is a drifting vehicle mix, not shown here to be a technology effect.
8. **Bimodality:** Car/Wagon/2-seater mean 240 (std 63) vs SUV/Truck/Van mean 271 (std 53). Body type is a plausible driver of the two bumps in the CO2 distribution; the KDE above shows how much it explains.
